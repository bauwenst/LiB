### Load the corpus

```python
import numpy as np

corpus_brtext = []
corpus_brtext_test = []
sents_set = set()

with open('br-text.txt') as f:
    for l in f.readlines():
        sents_set.add(l.replace('\n',''))
        
sents_set = list(sents_set)        
sents1 = []
for i in sents_set:
    sent = i.split(' ')
    sents1.append(sent)

for _ in range(400):
    corpus_brtext.append([])
    for i in set(np.random.choice(range(len(sents1)),200)):
        corpus_brtext[-1].append(sents1[i])
    corpus_brtext[-1] = [[''.join(j) for j in corpus_brtext[-1]], corpus_brtext[-1]]

sents2 = []
for i in sents_set[int(len(sents_set)*0.9):]:
    sent = i.split(' ')
    sents2.append(sent)
sents2 = [[''.join(j) for j in sents2], sents2]
corpus_brtext_test.append(sents2)
```

### Run LiB

```python
from libtok.vocabularisation import LiBVocabulariser
from libtok.structures import TrieList, SkipGram

model = LiBVocabulariser(
    probation_time= 10,
    max_len = 12,
    memory_in = 0.25,
    memory_out = 0.0001,
    update_rate = 0.2,

    mini_gap = 7,
    do_skipgrams=True
)

corpus_train = corpus_brtext
corpus_test = corpus_brtext_test

model.initialise(corpus_train[0][0]) # init the Lexicon memory with some unigrams in corpus

for epoch_id in range(5001):
    model.training_step(epoch_id, corpus_train, corpus_test)
```

```python
for k in model.memory[:200]:
    if isinstance(k, SkipGram):
        print(k.left, k.right)
```

```python
for k in model.memory:
    if k in model.memory.skipgram_frames:
        print(k, model.memory.skipgram_frames[k])
```

### See the chunk segmentation result and the subchunk segmentation result 

```python
from libtok.visualisation import LiBVisualiser

visualiser = LiBVisualiser(model.segmenter, model.memory)

article, article_raw = corpus_train[2]
onset, end = 10, 20
print('---\nchunks\n---')
visualiser.show_result(article_raw[onset:end], article[onset:end], decompose=False)
print('---\nsubchunks\n---')
visualiser.show_result(article_raw[onset:end], article[onset:end], decompose=True)
```
